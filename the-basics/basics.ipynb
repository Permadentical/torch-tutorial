# Basics

Let's get some basic ideas out of the way

Probably the easiest way to setup the device variable.

This basically points to what torch will use to perform its operations moving forward. This tutorial has pretty innocent operations, but in the future, it will start mattering.

(I use CPU throughout)

````python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
````

````python
tensor = torch.tensor([[1,2,1], [9,2,1]], dtype=torch.float16, 
                      requires_grad=True, device=device)
````

A tensor, to my understanding, is simply a **higher order matrix**, typically much higher in dimensions for us CS folks. A vector being 1-D, and a matrix being 2-D might be trivial versions of a matrix.

Why its **DIFFERENT** than puny vectors and tensors is because it assumes **multilinear relationships**

Great Source: [What is a tensor?](https://www.youtube.com/watch?v=bpG3gqDM80w)

````python
# Some basic what-is-thats

print(tensor)
print(tensor.dtype) # Simply checks the type of tensor
print(tensor.requires_grad) # This is to track the tensor. More later.
print(tensor.shape) # The... shape, like a matrix, (rows, cols)
````

```requires_grad``` basically says this:

- I see that you have a tensor (variable)
- I see that you want me to track this
- If I track this, I will keep knowledge of its changes
- I can find its gradient
- When you want that, it will be available

If you build a simple neural network, you need to keep track of certain values to change them (weights during backpropagation for example). This helps us keep track of the things we want to probably CHANGE somehow

Some basic **init** methods

````python
print(torch.eye(5)) # Creates an identity matrix
print(torch.ones([3,3])) # fills a nxm matrix with ones
print(torch.zeros([3,4]))# fills a nxm matrix with zeros
print(torch.rand([2,2])) # random numbers between 0 and 1 of some size you pass
print(torch.arange(0, 5)) # a for-loop-esque tensor with end not included
````

We can convert any py array type, or numpy array type to a tensor. This is usually done before processing anything for more efficient code

````python
my_list = [i for i in range(6)]
print(torch.tensor(my_list))
````

### Arithmetic!

````python
# Addition
x = torch.arange(0,5)
y = torch.arange(5,10)
z = x + y
print(z)

## You can be fancy and do:
z = torch.add(x, y)
print(z)

## add supports an alpha param though. This basically does x + a*y
z = torch.add(x, y, alpha=10)
z
````

````python
# Subtraction
x = torch.arange(0,5)
y = torch.arange(5,10)
z = y - x
print(z)

## You can be fancy and do:
z = torch.subtract(y, x)
print(z)

## subtract supports an alpha param as well.
z = torch.subtract(x, y, alpha=10)
z
````

````python
# Multiplication
x = torch.arange(0,5)
y = torch.arange(5,10)
z = y * x
print(f"{x}, \n{y}, \n\t\t= \n{z}\n----------")

## You can do a matmul
z = x @ y
print(z) 

# Dot product is different. It will be:
mat1 = torch.rand(3)
mat2 = torch.rand(3)

print(torch.dot(mat1, mat2))
````

````python
# Division
# Multiplication
x = torch.arange(1,6)
y = torch.arange(5,10)
z = y / x 

print(f"{x}, \n{y}, \n\t\t= \n{z}\n----------")

## You can do this as well
z = torch.div(y, x)
print(z) 
````

You will notice is casts the type automatically in the first division. Tensors are also supposed to be the same type. (Unlike python lists, ew)

````python
# Exponents!
z = torch.pow(x, 2)
print(z)

# Or
z = x.pow(2)
z
````

````python
# you can also do comparisons
print(y > x)
print(torch.any(y < x)) # Any checks if ANYTHING in the array computes to True.
x.equal(y)
````

What happens if I want to work with a diverse set of tensors?

````python
mat1 = torch.rand([1, 5])
mat2 = torch.rand([10, 5])

print(mat1 - mat2)
````

Why does this work?

because of **Broadcasting!**

Broadcasting basically applies the 1-D tensor over each dimension (row, if you're confused) for the 10x5 tensor.

[Good Source](https://numpy.org/doc/stable/user/basics.broadcasting.html)

Other useful things I found at Torch HQ

````python
# You can shortcut your way to lots of things

# sum!
x = torch.arange(0,5)
y = torch.arange(10,15)
print(torch.sum(x, dtype=torch.int16))

# max, min
values, idx = torch.max(x, dim=0) 
# ^ dim is important here. it specifies where you want the max
print(values, idx)
values, idx = torch.min(y, dim=0) 
print(values, idx)

# argmax, argmin: gives you the idx directly
print(torch.argmax(x, dim=0))
````
